```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

data = {
    'Income': [25000, 50000, 30000, 70000, 60000,
               20000, 80000, 45000],
    'CreditHistory': ['Good', 'Good', 'Poor', 'Good',
                      'Good', 'Poor', 'Good', 'Poor'],
    'Employment': ['Yes', 'Yes', 'No', 'Yes',
                   'Yes', 'No', 'Yes', 'No'],
    'LoanApproved': ['No', 'Yes', 'No', 'Yes',
                     'Yes', 'No', 'Yes', 'No']
}

df = pd.DataFrame(data)

le = LabelEncoder()

for col in ['CreditHistory', 'Employment', 'LoanApproved']:
    df[col] = le.fit_transform(df[col])

X = df[['Income', 'CreditHistory', 'Employment']]
y = df['LoanApproved']

model = GaussianNB()
model.fit(X, y)

pred = model.predict(X)

print("Predicted Loan Status:", pred)
print("Actual Loan Status:", y.values)
print("Accuracy:", accuracy_score(y, pred))

sample = [[55000, 1, 1]]
result = model.predict(sample)[0]

print("Loan Status:",
      "Approved" if result == 1 else "Rejected")
```
