```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import LabelEncoder

data = {
    'Year': [2015, 2016, 2017, 2018, 2019,
             2020, 2021, 2022],
    'Mileage': [50000, 45000, 40000, 35000,
                30000, 25000, 20000, 15000],
    'Brand': ['Honda', 'Toyota', 'Honda', 'Toyota',
              'Honda', 'Toyota', 'Honda', 'Toyota'],
    'Price': [400000, 450000, 500000, 550000,
              600000, 650000, 700000, 750000]
}

df = pd.DataFrame(data)

le = LabelEncoder()
df['Brand'] = le.fit_transform(df['Brand'])

X = df[['Year', 'Mileage', 'Brand']]
y = df['Price']

model = LinearRegression()
model.fit(X, y)

pred = model.predict(X)

print("Actual Prices:", list(y))
print("Predicted Prices:", pred)

new_car = [[2023, 10000, le.transform(['Honda'])[0]]]
print("Estimated Car Price:", model.predict(new_car)[0])
```
